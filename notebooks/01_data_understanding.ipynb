## 3. Column Names

We inspect all available features and identify the target variable.

## 4. Data Types

Understanding data types helps us distinguish numerical and categorical features.

```python
df.dtypes
```

```python
df.info()
```

```python
numeric_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns
```

```python
print("Numerical columns:")
print(numeric_columns.tolist())

print("\nCategorical columns:")
print(categorical_columns.tolist())
```

## 5. Missing Values

We identify columns containing missing values and measure their amount.

Missing values must be analyzed before deciding how they should be handled.

```python
missing_count = df.isnull().sum()

missing_count = missing_count[missing_count > 0]

missing_count.sort_values(ascending=False)
```

```python
missing_percentage = (
    df.isnull().mean() * 100
).sort_values(ascending=False)

missing_percentage[missing_percentage > 0]
```

```python
missing_df = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percentage": df.isnull().mean() * 100
})

missing_df = missing_df[
    missing_df["missing_count"] > 0
].sort_values(
    "missing_percentage",
    ascending=False
)

missing_df
```

## 6. Duplicate Records

We check whether the dataset contains duplicated rows.

```python
duplicate_count = df.duplicated().sum()

print(f"Duplicate rows: {duplicate_count}")
```

```python
df[df.duplicated()].head()
```

```python
for column in categorical_columns:
    print(f"\n{'=' * 60}")
    print(column)
    print(f"Unique values: {df[column].nunique()}")
    print(df[column].value_counts(dropna=False).head(10))
```

## 7. Numerical Feature Statistics

We inspect descriptive statistics for numerical variables.

```python
df[numeric_columns].describe().T
```

```python
TARGET = "Converted"
print(df[TARGET].value_counts())
```

```python
print(df[TARGET].value_counts(normalize=True) * 100)
```

## 8. Target Variable

The target variable represents whether a lead converted.

We inspect the class distribution to understand whether the target is balanced or imbalanced.

```python
target_distribution = df[TARGET].value_counts().to_frame("count")

target_distribution["percentage"] = (
    target_distribution["count"]
    / len(df)
    * 100
)

target_distribution
```

```python
print("=== MISSING VALUES ===")
print(missing_df)

print("\n=== DUPLICATES ===")
print(df.duplicated().sum())

print("\n=== TARGET COUNT ===")
print(df["Converted"].value_counts())

print("\n=== TARGET PERCENTAGE ===")
print(df["Converted"].value_counts(normalize=True) * 100)
```
